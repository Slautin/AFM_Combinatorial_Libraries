```python
import shutil
from afm_lib.config import RUNS_ROOT

freed = 0
for run in sorted(RUNS_ROOT.iterdir()):
    c = run / "cache"
    if not c.is_dir():
        continue
    n = sum(f.stat().st_size for f in c.rglob("*") if f.is_file())
    shutil.rmtree(c)
    freed += n
    print(f"{run.name}: {n/1e6:8.1f} MB")
print(f"\nfreed {freed/1e6:.0f} MB")
```
