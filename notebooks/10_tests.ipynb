```python
from pathlib import Path
from afm_lib.schemas.recipe import load_recipe, site_program

```

```python
for p in sorted(Path("../recipes").glob("*.yaml")):          # every old recipe still loads
    r = load_recipe(p)
    kinds = {m.kind for m in site_program(r)}
    print(f"{p.name:32s} mode={r.mode!r:22s} sites={len(r.sites):3d} program={kinds}")

r = load_recipe("../recipes/smoke_2site_ac.yaml")
m = site_program(r)
assert r.mode == "AC Air Topography" and len(m) == 1 and m[0].kind == "scan"
assert m[0].scan_settings.pixels == 64 and m[0].loop_settings is None
print("ok:", m[0])
```
